```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error,r2_score
```

```python
df = pd.read_csv("temperature.csv")
df.head()
```

```python
df.shape 
```

```python
df.columns
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.corr(numeric_only=True)
```

```python
X = df[["BASEL_temp_min"]]
y = df["BASEL_temp_mean"]

X.shape
```

```python
X.head()
```

```python
y.head()
```

```python
plt.figure(figsize = (10,6))
plt.scatter(X, y, color = "blue")

plt.xlabel("Minimum Temperature", color = "red")
plt.ylabel("Mean Temperature", color = "red")

plt.title("Minimum Temperature VS Mean Temperature", color = "red")

plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split (X, y, random_state = 42, test_size = 0.2)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
poly = PolynomialFeatures(degree = 2)
X_train_poly = poly.fit_transform(X_train)
```

```python
X_train_poly.shape
```

```python
X_test_poly = poly.transform(X_test)
```

```python
X_test_poly.shape
```

```python
model = LinearRegression()
```

```python
model.fit(X_train_poly, y_train)
```

```python
y_pred = model.predict(X_test_poly)
y_pred[:5]
```

```python
comparision = pd.DataFrame({
    "Actual Temperature": y_test.values,
    "Predicted Temperature":  y_pred
})

comparision.head(10)
```

```python
mae = mean_absolute_error(y_test, y_pred)

print("MAE:", mae)
```

```python
mse = mean_squared_error(y_test, y_pred)
print("MSE:", mse)
```

```python
rmse = np.sqrt(mse)

print("RMSE:", rmse)
```

```python
r2 = r2_score(y_test, y_pred)

print("R² Score:", r2)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(y_test, y_pred)

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)

plt.xlabel("Actual Mean Temperature", color = "red")
plt.ylabel("Predicted Mean Temperature", color = "red")
plt.title("Actual vs Predicted Mean Temperature", color = "red")

plt.show()
```

```python
X_sorted = np.sort(X.values, axis=0)

X_sorted_poly = poly.transform(X_sorted)

y_sorted_pred = model.predict(X_sorted_poly)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(X, y)
plt.plot(X_sorted, y_sorted_pred)

plt.xlabel("Minimum Temperature")
plt.ylabel("Mean Temperature")
plt.title("Polynomial Regression Curve")

plt.show()
```

```python
print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)
```

## Conclusion

In this project, I used Polynomial Regression to predict the mean temperature of Basel using the minimum temperature.

I first transformed the input feature using PolynomialFeatures with degree 2, which created the features X and X². Then, I trained a Linear Regression model on these transformed features.

The model was evaluated using MAE, MSE, RMSE, and R² Score. I also compared actual and predicted temperatures and visualized the polynomial regression curve.

This project helped me understand how Polynomial Regression can model a curved relationship between an input feature and a target variable.
